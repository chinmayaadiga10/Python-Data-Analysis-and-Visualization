```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
array1=np.array([1,2,3,4,5])
array2=np.random.random((3,4))
array3=np.zeros((3,3))
```

### Saving arrays as flat files and loading them


```python
np.save('array1.npy',array1)
np.save('array2.npy',array2)
np.save('array3.npy',array3)
```

#### Loading the saved arrays 

```python
loaded_array1=np.load('array1.npy')
print(loaded_array1)
```

### Loading and displaying images using numpy and matplotlib

```python
try:
    logo=np.load('numpy-logo.npy')
except FileNotFoundError:
    print("Numpy logo not found")
else:
      # Displaying the loaded image using matplotlib =>
    plt.figure(figsize=(10,5))
    plt.title("Numpy Logo")
    plt.subplot(121)
    plt.imshow(logo)
    plt.grid(False)
    
    dark_logo=1-logo
    plt.subplot(122)
    plt.imshow(dark_logo)
    plt.title("Dark Logo")
    plt.grid(False)
```
