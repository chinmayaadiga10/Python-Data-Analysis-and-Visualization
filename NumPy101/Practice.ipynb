### Check Valid Sudoku  

```python
sudoku= np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
np.sum(sudoku,axis=1)
```

```python
np.sum(sudoku,axis=0)
```

```python
valid=False
rows=np.sum(sudoku,axis=0)
columns=np.sum(sudoku,axis=0)
for row_sum in rows:
    if row_sum!=45:
        break
else:
    valid=True
    
for column_sum in columns:
    if column_sum!=45:
        break
else:
    valid=True
    
if valid:
    print("Sudoku is valid")
else:
    print("Sudoku is invalid")
    
```

### Analysis on Student Data

```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
average_age=data[:,0].mean()
print(average_age)
```

```python
#Extract Math marks of all students.
math_marks=data[:,1]
print(math_marks)
```

```python
#Find the highest Science mark.
science_highest=data[:,2].max()
print(science_highest)
```

```python
#Get details of the student who scored more than 90 in Math.
condition=data[:,1]>90
data[condition]
```

```python
#Increase Math marks of all students by 5.
data[:,1]+10
```

```python
#Find how many students are younger than 19.
data[data[:,0]<19]
no_of_students=np.sum(data[:,0]<19)
print(no_of_students)
```

```python
#Calculate the average marks in each subject (column-wise mean).
average_marks=np.mean(data[:,1:],axis=0)
average_marks
```

```python
#Get data of students who scored at least 80 in both subjects.
data[(data[:,1]>=80)&(data[:,2]>=80)]
```
