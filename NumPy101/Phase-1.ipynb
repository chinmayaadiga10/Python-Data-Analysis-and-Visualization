#### Understanding the basics of NumPy

```python
import numpy as np
```

```python
# OD Vector ->
arr=np.array(42)
print(arr)

# 1D array ->
array_1d=np.array([1,2,3,4,5])
print(f"1D array is : {array_1d}")

# 2D array ->
array_2d=np.array([[1,2,3,4],[5,6,7,8]])
print(f"2D array is : \n{array_2d}")
```

#### List vs NumPy array 

```python
python_list=[1,2,3,4,5]
print(f"Python list multiplication : {python_list*2}")

np_array=np.array([1,2,3,4,5])
print(f"Np array multiplication: {np_array*2}")
```

Comparing time taken by normal multiplication and numpy multiplication

```python
import time

start=time.time()
python_list=[i*2 for i in range(1000000)]
print(f"Python list multiplication time taken  : {time.time()-start}")


start=time.time()
np_array=np.arange(1000000)*2
print(f"Numpy  multiplication time taken  : {time.time()-start}")
```

##### Creating array from scratch in NumPy

```python
#using the zeros and ones method allows us to create arrays with all values equal to either 0/1
zeros=np.zeros((3,4))
print("zeroes array :\n",zeros)

ones=np.ones((2,3))
print("ones array :\n",ones)

# creating array of constant values -> 

full=np.full((2,2),7)
print("constant array :\n",full)


# creating a array of random values -> 

random=np.random.random((2,3))
print("random array :\n",random)

#creating a sequence 
# range -> non inclusive outer boundary
sequence=np.arange(start=0,step=2,stop=10)
print("sequence array :\n",sequence)
```

#### Array Properties

```python
arr=np.array([[1,2,3],
              [4,5,6]])
print(f"Shape of array is : {arr.shape}")
print(f"Size of array is : {arr.size}")
print(f"Dimension of array is : {arr.ndim}")
print(f"Data type of array is : {arr.dtype}")

#even if 1 element of the array is float, then the data type is shown as float
```

Array reshaping operations

```python
arr=np.arange(12)
print(f"Original array is : {arr}")

reshaped=arr.reshape((3,4))
print(f"Reshaped array is :\n {reshaped}")

flattened=reshaped.flatten()
print(f"Flattened array is :\n {flattened}")

raveled=reshaped.ravel()
print(f"Raveled array is :\n {raveled}")

# the main difference between raveled and flattened is that raveled makes changes in the original array while the raveled returns a copy of the flattened


# Finding transpose of matrix -> 

transpose=reshaped.T
print(f"The transpose array is : \n {transpose}")
```

#### Vectors, Matrix and Tensors in NumPy

```python
vector=np.array([1,2,3,4,5])
print(" Vector :\n",vector)

matrix=np.array([[1,2,3],[4,5,6]])
print("Matrix :\n",matrix)

tensor=np.array([[[1,2],[3,4]],
                 [[5,6],[7,8]]])
print("Tensor is  :\n",tensor)
```
