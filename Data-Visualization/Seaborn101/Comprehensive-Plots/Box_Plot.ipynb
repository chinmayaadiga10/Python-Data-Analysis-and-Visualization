## Box Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df=sns.load_dataset("tips")
df
```

```python
sns.set(style="whitegrid")
```

### Basic Syntax

```python
sns.boxplot(x="day",y="total_bill",data=df)
plt.show()
```

### Adding differentiator column

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="time")
plt.show()
```

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="sex")
plt.show()
```

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="sex",color="green")
plt.show()
```

### Changing order in x-axis

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="sex",color="orange",order=['Fri','Sun','Thur','Sat'])
plt.show()
```

### Displaying means

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="sex",showmeans=True)
plt.show()
```

### Modifying linewidth, palette parameters

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="sex",showmeans=True,linewidth=3)
plt.show()
```

```python
sns.boxplot(x="day",y="total_bill",data=df,hue="sex",showmeans=True,linewidth=3,palette="plasma")
plt.show()
```

### Changing Orientation

```python
sns.boxplot(x="total_bill",y="day",data=df,hue="sex",showmeans=True,linewidth=3,palette="plasma",orient="h")
plt.show()
```

### Box Plot of single Value/Column

```python
sns.boxplot(y=df['total_bill'])
plt.show()
```

```python
sns.boxplot(x=df['total_bill'])
plt.show()
```
