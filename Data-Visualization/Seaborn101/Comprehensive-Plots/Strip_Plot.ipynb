## Strip Plot
#### Basically a scatter plot that differentiates different categories

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

## Basic Syntax

```python
sns.stripplot(x="day",y="total_bill",data=df)
plt.show()
```

### Adding a differentiator column using hue function

```python
sns.stripplot(x="day",y="total_bill",data=df,hue="sex")
plt.show()
```

### Adding color to the graph using palette and color parameters

```python
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r")
plt.show()
```

### Changing Linewidth and Edgecolor

```python
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",linewidth=1)
plt.show()
```

```python
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",linewidth=1,edgecolor="red")
plt.show()
```

### Modifying the size of individual observation and shape using marker

```python
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",linewidth=1,edgecolor="red",size=9)
plt.show()
```

```python
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",linewidth=1,edgecolor="red",marker="d")
plt.show()
```

### Changing opacity using alpha parameters

```python
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",linewidth=1,edgecolor="red",marker="d",alpha=0.6)
plt.show()
```

### Strip Plot of a single value

```python
sns.stripplot(x=df["total_bill"])
plt.show()
```

```python
sns.stripplot(y=df["total_bill"])
plt.show()
```
