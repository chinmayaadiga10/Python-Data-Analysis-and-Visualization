## Histogram in Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("penguins")
df
```

### Basic Syntax

```python
sns.histplot(df['bill_length_mm'])
plt.show()
```

### Bins and KDE parameters

#### KDE stands for Kernel Density Estimator

```python
sns.histplot(df['flipper_length_mm'])
plt.show()
```

```python
sns.histplot(df['flipper_length_mm'],bins=[170,180,190,200,210,220,230,240])
plt.show()
```

```python
sns.histplot(df['flipper_length_mm'],bins=[170,180,190,200,210,220,230,240],kde=True)
plt.show()
```

```python
sns.histplot(df['flipper_length_mm'],kde=True,color="orange")
plt.show()
```
