```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("penguins")
df
```

```python
sns.barplot(x=df['island'],y=df['bill_length_mm'])
plt.show()
```

```python
sns.barplot(x="island",y="bill_length_mm",data=df)
plt.show()
```

```python
sns.barplot(x="island",y="bill_length_mm",data=df,hue="sex")
plt.show()
```

### Changing order of representation

```python
orders=["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders)
plt.show()
```

### Changing Hue Order

```python
orders=["Dream","Torgersen","Biscoe"]
hue_order=["Female","Male"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders,hue="sex",hue_order=hue_order)
plt.show()
```

### CI Parameter - Confidence Interval

```python
orders=["Dream","Torgersen","Biscoe"]
hue_order=["Female","Male"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders,hue="sex",hue_order=hue_order,ci=80)
plt.show()
```

### Changing Orientation of Graph

```python
sns.barplot(x="bill_depth_mm",y="bill_length_mm",data=df,orient="v")
plt.show()
```

```python
sns.barplot(x="bill_depth_mm",y="bill_length_mm",data=df,orient="h")
plt.show()
```

### Color and Palette Parameters

```python
orders=["Dream","Torgersen","Biscoe"]
hue_order=["Female","Male"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders,hue="sex",hue_order=hue_order,color="orange")
plt.show()
```

```python
orders=["Dream","Torgersen","Biscoe"]
hue_order=["Female","Male"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders,hue="sex",hue_order=hue_order,palette="icefire")
plt.show()
```

### Saturation and Alpha Parameters

```python
orders=["Dream","Torgersen","Biscoe"]
hue_order=["Female","Male"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders,hue="sex",hue_order=hue_order,saturation=20)
plt.show()
```

```python
orders=["Dream","Torgersen","Biscoe"]
hue_order=["Female","Male"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders,hue="sex",hue_order=hue_order,saturation=20,alpha=0.5)
plt.show()
```

### Setting dark grid

```python
sns.set(style="dark")
orders=["Dream","Torgersen","Biscoe"]
hue_order=["Female","Male"]
sns.barplot(x="island",y="bill_length_mm",data=df,order=orders,hue="sex",hue_order=hue_order)
plt.show()
```
