## Count Plot in Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

### Basic Syntax

#### Count plot shows the total no.of visitors

```python
sns.countplot(x="sex",data=df)
plt.show()
```

#### Bar Plot shows the mean value

```python
sns.barplot(x="sex",y="size",data=df)
plt.show()
```

### Categorising the tips given by smokers and non-smokers

```python
sns.countplot(x="sex",data=df,hue="smoker")
plt.show()
```

### Changing orientation of graph to horizontal

```python
sns.countplot(y="sex",data=df,hue="smoker")
plt.show()
```

### Changing color using color list

```python
sns.countplot(x="sex",data=df,hue="smoker",palette="prism")
plt.show()
```

### Changing saturation and adding custom color to the graph

```python
sns.countplot(x="sex",data=df,hue="smoker",color="violet",saturation=0.2)
plt.show()
```
