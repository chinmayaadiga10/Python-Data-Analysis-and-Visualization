```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

### Basic Syntax

```python
sns.catplot(x="tip",y="size",data=df)
plt.show()
```

```python
sns.catplot(x="tip",y="size",data=df,hue="sex")
plt.show()
```

```python
sns.catplot(x="tip",y="size",data=df,hue="sex",palette="winter_r")
plt.show()
```

```python
sns.catplot(x="tip",y="size",data=df,hue="sex",palette="winter_r",height=6)
plt.show()
```

```python
sns.catplot(x="tip",y="size",data=df,hue="sex",palette="winter_r",kind="bar")
plt.show()
```

```python
sns.catplot(x="tip",y="size",data=df,hue="sex",palette="winter_r",kind="box")
plt.show()
```
