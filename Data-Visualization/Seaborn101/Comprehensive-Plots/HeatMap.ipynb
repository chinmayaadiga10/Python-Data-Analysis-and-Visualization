## HeatMaps in Seaborn

#### Used extensively in machine learning

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=np.linspace(1,10,20).reshape(4,5)
data
```

### Basic Syntax

```python
sns.heatmap(data)
plt.show()
```

### Working with CSV data

```python
df=sns.load_dataset("anagrams")
df
```

```python
x=df.drop(columns=['attnr'],axis=1)
x
```

```python
sns.heatmap(x)
plt.show()
```

### Altering minimum and maximum values in heatmap

```python
sns.heatmap(data,vmin=0,vmax=12)
plt.show()
```

### Changing colors of heatmap

```python
sns.heatmap(data,vmin=0,vmax=20,cmap="gist_heat")
plt.show()
```

### Displaying individual values

```python
sns.heatmap(data,vmin=0,vmax=20,cmap="gist_heat",annot=True)
plt.show()
```

```python
z=np.linspace(1,10,10).reshape(2,5)
z
```

```python
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=True)
plt.show()
```

### Replacing values with text

```python
values=np.array([["a0","a1","a2","a3","a4"],["b0","b1","b2","b3","b4"]])
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=values,fmt="s")
plt.show()
```

### Making modifications with dictionary

```python
changes={"fontsize":12,"color":"orange"}
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=values,fmt="s",annot_kws=changes)
plt.show()
```

### Linewidth property

```python
changes={"fontsize":12,"color":"orange"}
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=values,fmt="s",annot_kws=changes,linewidth=10)
plt.show()
```

```python
changes={"fontsize":12,"color":"orange"}
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=values,fmt="s",annot_kws=changes,linewidth=10,linecolor="black")
plt.show()
```

### Removing colorbar and labels

```python
changes={"fontsize":12,"color":"orange"}
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=values,fmt="s",annot_kws=changes,linewidth=10,cbar=False)
plt.show()
```

```python
changes={"fontsize":12,"color":"orange"}
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=values,fmt="s",annot_kws=changes,linewidth=10,xticklabels=False)
plt.show()
```

```python
changes={"fontsize":12,"color":"orange"}
sns.heatmap(z,vmin=0,vmax=20,cmap="gist_heat",annot=values,fmt="s",annot_kws=changes,linewidth=10,xticklabels=False,yticklabels=False)
plt.show()
```
