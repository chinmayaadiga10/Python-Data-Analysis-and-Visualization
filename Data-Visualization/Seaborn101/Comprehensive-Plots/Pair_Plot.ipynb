## Pair Plots in Seaborn

#### Used for comparison between graphs

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

### Basic Syntax

```python
sns.pairplot(df)
plt.show()
```

### Differentiating graph

```python
sns.pairplot(df,hue="sex")
plt.show()
```

### Selecting only particular columns from the dataset

```python
sns.pairplot(df,vars=["total_bill","tip"],hue="sex")
plt.show()
```

### Changing Hue Order

```python
sns.pairplot(df,vars=["total_bill","tip"],hue="sex",hue_order=["Female","Male"])
plt.show()
```

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],x_vars=["total_bill","tip"])
plt.show()
```

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],y_vars=["total_bill","tip"])
plt.show()
```

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"])
plt.show()
```

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],kind="reg")
plt.show()
```

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],kind="hist")
plt.show()
```

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],kind="scatter")
plt.show()
```

### Changing Markers

```python
sns.pairplot(df,hue="sex",hue_order=["Female","Male"],markers=["*","^"])
plt.show()
```
