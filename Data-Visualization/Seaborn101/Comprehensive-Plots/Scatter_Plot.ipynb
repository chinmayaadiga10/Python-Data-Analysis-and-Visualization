## Scatter Plot in Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("penguins")
df
```

### Basic Syntax

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df)
plt.show()
```

### Showing differentiation in data with a differentiating column

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex")
plt.show()
```

### Modifying graph with style and size parameters

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex")
plt.show()
```

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",size="sex")
plt.show()
```

#### Passing custom size values

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",size="sex",sizes=(70,50))
plt.show()
```

### Playing with palettes parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",palette="gist_rainbow")
plt.show()
```

### Modifying opacity using alpha parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",palette="Accent",alpha=0.7)
plt.show()
```

### Making changes in the marker

```python
m={"Male":"*","Female":"o"}
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",palette="gist_rainbow",markers=m)
plt.show()
```
