```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

```python
sns.boxplot(y=df['total_bill'])
plt.show()
```

```python
sns.barplot(x=df['day'],y=df['total_bill'])
plt.grid()
plt.show()
```

```python
sns.set_style("dark")
sns.barplot(x=df['day'],y=df['total_bill'])
plt.show()
```

```python
sns.set_style("whitegrid")
sns.barplot(x=df['day'],y=df['total_bill'])
plt.show()
```

```python
sns.set_style("darkgrid")
sns.barplot(x=df['day'],y=df['total_bill'])
plt.show()
```

```python
sns.set_style("ticks")
sns.barplot(x=df['day'],y=df['total_bill'])
plt.show()
```

### Removing axes line

```python
sns.barplot(x=df['day'],y=df['total_bill'])
sns.despine()
plt.show()
```

### Scale and context properties

```python
plt.figure(figsize=(10,5))
sns.barplot(x=df['day'],y=df['total_bill'])
sns.despine()
plt.show()
```

### Font Properties

```python
sns.set_context("poster",font_scale=1)
sns.barplot(x=df['day'],y=df['total_bill'])
plt.show()
```

```python
sns.set_context("paper",font_scale=1)
sns.barplot(x=df['day'],y=df['total_bill'])
plt.show()
```

### Color Parameters

```python

sns.barplot(x=df['day'],y=df['total_bill'],palette="Reds_r")
plt.show()
```
