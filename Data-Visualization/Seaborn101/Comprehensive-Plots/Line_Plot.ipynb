## Line Plots in Seaborn

#### Unlike matplotlib, seaborn plots only work when used on a dataset and not otherwise

#### All matplotlib functions can be applied here

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
var=[1,2,3,4,5,6,7]
var_1=[3,1,5,6,2,7,4]

plt.plot(var,var_1)
plt.show()
```

### Creating a dataframe

```python
var=[1,2,3,4,5,6,7]
var_1=[3,1,5,6,2,7,4]

df=pd.DataFrame({"var":var,"var_1":var_1})

sns.lineplot(x="var",y="var_1",data=df)
plt.show()
```

### Using Seaborn without creating Dataframe

```python
var=[1,2,3,4,5,6,7]
var_1=[3,1,5,6,2,7,4]

sns.lineplot(x=var,y=var_1)
plt.show()
```

```python
df=sns.load_dataset("penguins")
df
```

```python
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=df)
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df)
plt.show()
```

### Creating a comparison graph with respect to another data point

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex")
plt.show()
```

#### Labels and legends are added automatically to the graph in SeaBorn

### Size and style parameters in Seaborn

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",size=40)
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex")
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",palette="Accent")
plt.show()
```

### Adding markers to the graph

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",markers=["o",">"])
plt.show()
```

### Removing legend using parameters and adding dashed lines

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",markers=["o",">"],dashes=False,legend=False)
plt.show()
```

### Adding Grid and Title using Seaborn

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",markers=["o",">"],dashes=True)

plt.title("Seaborn Graph")
plt.grid()
plt.show()
```
