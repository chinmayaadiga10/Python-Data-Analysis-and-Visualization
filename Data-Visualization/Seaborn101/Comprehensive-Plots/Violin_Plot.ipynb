## Violin Plots in Seaborn

#### These are similar to box and whisker plots in Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

### Basic Syntax

```python
sns.violinplot(x="day",y="total_bill",data=df)
plt.show()
```

### Making use of differentiators

```python
sns.violinplot(x="day",y="total_bill",data=df,hue="time")
plt.show()
```

### Linewidth and palette parameters

```python
sns.violinplot(x="day",y="total_bill",data=df,linewidth=2)
plt.show()
```

```python
sns.violinplot(x="day",y="total_bill",data=df,linewidth=1,palette="Pastel1_r")
plt.show()
```

### Using order parameter

```python
sns.violinplot(x="time",y="total_bill",data=df,linewidth=1,order=['Dinner','Lunch'])
plt.show()
```

### Saturation and Color Parameters

```python
sns.violinplot(x="time",y="total_bill",data=df,linewidth=1,order=['Dinner','Lunch'],saturation=0.2)
plt.show()
```

```python
sns.violinplot(x="time",y="total_bill",data=df,linewidth=1,order=['Dinner','Lunch'],color="yellow")
plt.show()
```

### Split and Scale Parameters

```python
sns.violinplot(x="day",y="total_bill",data=df,hue="sex",split=True)
plt.show()
```

```python
sns.violinplot(x="day",y="total_bill",data=df,hue="sex",split=True,scale="count")
plt.show()
```

### Creating a single violin Plot

```python
sns.violinplot(df["total_bill"])
plt.show()
```

```python
sns.violinplot(x=df["total_bill"])
plt.show()
```
