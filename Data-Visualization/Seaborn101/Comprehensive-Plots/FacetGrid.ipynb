## Facet Grid - creating Multi Plots in Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset("tips")
df
```

## Basic Syntax

```python
fg=sns.FacetGrid(df)
fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

### Division on basis of Gender

```python
fg=sns.FacetGrid(df,col="sex")
fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

```python
fg=sns.FacetGrid(df,col="sex",hue="day")
fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

```python
fg=sns.FacetGrid(df,col="sex",hue="day")
fg.map(plt.scatter,"total_bill","tip").add_legend()
plt.show()
```

```python
fg=sns.FacetGrid(df,col="day",hue="sex")
fg.map(plt.scatter,"total_bill","tip")
plt.show()
```

### Creating bar plots

```python
fg=sns.FacetGrid(df,col="day",hue="sex")
fg.map(plt.bar,"total_bill","tip")
plt.show()
```
