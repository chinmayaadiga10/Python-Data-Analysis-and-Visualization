# Regression Plots

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
df
```

```python
sns.lmplot(x='total_bill',y='tip',data=df)
plt.show()
```

```python
sns.lmplot(x='total_bill',y='tip',data=df,hue="sex")
plt.show()
```

```python
sns.lmplot(x='total_bill',y='tip',data=df,palette="rainbow")
plt.show()
```

```python
sns.lmplot(x='total_bill',y='tip',data=df,hue="sex",markers=['d','*'])
```
