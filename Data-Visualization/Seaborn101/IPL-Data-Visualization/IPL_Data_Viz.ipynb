### Finding the team with the most victories

```python
df['match_winner'].value_counts().max()
```

```python
match_wins=df['match_winner'].value_counts()
```

```python
sns.barplot(y=match_wins.index,x=match_wins.values,palette="rocket")
plt.title("No.of matches won by teams")
plt.show()
```

### Finding toss decision trends

```python
sns.countplot(x=df['toss_decision'],palette="inferno")
plt.show()
```

## Key Player Performances

### Finding the player with the most man of the match performances

```python
df['player_of_the_match'].value_counts().head(10)
```

```python
award_count=df['player_of_the_match'].value_counts().head(10)
```

```python
award_count.values
```

```python
award_count.index
```

```python
sns.barplot(y=award_count.index,x=award_count.values,palette="rocket")
plt.title("Man of the Match Awards")
plt.show()
```

### Top 2 highest scorers

```python
df['top_scorer'].value_counts().head(2)
```

```python
df.groupby(by='top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
```

```python
orange_cap=df.groupby(by='top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
orange_cap
```

```python
sns.barplot(x=orange_cap.index,y=orange_cap.values,palette="Oranges_r",saturation=2)
plt.title("Most runs scored")
plt.show()
```

### Top 10 best bowling figures

```python
df['best_bowling_figure']
```

```python
df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
```

```python
df['most_wickets']=df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
```

```python
type(df['most_wickets'][0])
```

```python
df['most_wickets']=df['most_wickets'].astype(int)
```

```python
type(df['most_wickets'][0])
```

```python
df.groupby(by="best_bowling")['most_wickets'].sum().sort_values(ascending=False).head(10)
```

```python
purple_cap=df.groupby(by="best_bowling")['most_wickets'].sum().sort_values(ascending=False).head(10)
```

```python
sns.barplot(x=purple_cap.values,y=purple_cap.index,palette="mako")
plt.show()
```
