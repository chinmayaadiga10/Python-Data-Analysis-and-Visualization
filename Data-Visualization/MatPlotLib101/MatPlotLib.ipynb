### Constructing Bar Graph

```python
# syntax -> plt.bar(x,height,color='color name',width=value,label='label name')
products=['A','B','C','D']
sales=[1000,1500,800,1500]

plt.bar(products,sales,color='red',label='2025 Sales')
plt.xlabel('Products')
plt.ylabel('Sales ')
plt.title('Product Sales Comparison')
plt.legend()
plt.show()
```

#### Horizontal Bar Charts

```python
products=['A','B','C','D']
sales=[1000,1500,800,1500]

plt.barh(products,sales,color='red',label='2025 Sales')
plt.ylabel('Products')
plt.xlabel('Sales ')
plt.title('Product Sales Comparison')
plt.legend()
plt.show()
```

### Constructing Pie Charts

```python
# Syntax -> plt.pie(values, label=label_list,colors='color_list',autopct='%1.1f%%')

regions=['North','South','East','West']
revenue=[3000,2000,1500,1000]

plt.pie(revenue,labels=regions,autopct='%1.1f%%')#we can specify by our own colors by passing a list of colors. in pie chart -> max 6 categories
plt.title('Revenue Contribution By Region')
plt.show()
```

### Constructing Histograms

Histograms are used for numerical continuous data

```python
# syntax -> plt.hist(data,bins=no.of bins, color='colorname', edgecolor='black')

scores=[45,67,89,56,78,88,92,60,74,81,59,66,75,82,90,85,70,73,68,77]
plt.hist(scores,bins=5,color='yellow',edgecolor='black')
plt.xlabel('Score Range')
plt.ylabel('No.of Students')
plt.title('Score Distribution Of Students')
plt.show()
```

### Constructing Scatter Plots

#### Scatter plots are graphs where each point represents one point with two values - used to determine the relationship between the two values

```python
#Syntax -> plt.scatter(x,y,color='color label', marker='marker style', label='label name - for the legend')

hours_studied=[1,2,3,4,5,6,7,8]
score=[50,55,60,65,70,75,80,85]

plt.scatter(hours_studied,score,color='green',marker='d',label='student data')
plt.xlabel('Hours Studied')
plt.ylabel('Marks Scored')
plt.title('Relationship Between Hours Studied and Marks Studied')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Syntax -> plt.scatter(x,y,color='color label', marker='marker style', label='label name - for the legend')



plt.scatter([1,2,3],[50,55,60],color='blue',marker='s',label='Class A')#group 1
plt.scatter([1,2,3],[47,60,69],color='orange',marker='^',label='Class B')#group 2
plt.xlabel('Hours Studied')
plt.ylabel('Marks Scored')
plt.title('Comparison of Two Classes')
plt.legend()
plt.grid(True)
plt.show()
```

## Understanding SubPlots in MatPlotLib

```python
#Syntax -> plt.subplot(nrows,ncols,index)

x=[1,2,3,4]
y=[10,20,15,25]

plt.subplot(1,2,1) # 1st row, 2nd column, index=1
plt.plot(x,y)
plt.title('Line Chart')

plt.subplot(1,2,2) # 1st row, 2nd column, index=2
plt.bar(x,y)
plt.title('Bar Chart')

plt.show()
```

### Using Professsional techniques to create subplots

```python
#Syntax -> fig,ax=plt.subplots(nrow,ncols,figsize=(width,height))

fig,ax=plt.subplots(1,2,figsize=(10,5))

x=[1,2,3,4]
y=[10,20,15,25]


ax[0].plot(x,y)
ax[0].set_title('Line Chart')

ax[1].bar(x,y)
ax[1].set_title("Bar Graph")

plt.show()
```

#### Making The Graphs appear in the entire canvas

```python
#Syntax -> fig,ax=plt.subplots(nrow,ncols,figsize=(width,height))

fig,ax=plt.subplots(1,2,figsize=(10,5))

x=[1,2,3,4]
y=[10,20,15,25]


ax[0].plot(x,y,color="blue")
ax[0].set_title('Line Chart')

ax[1].bar(x,y,color='yellow')
ax[1].set_title("Bar Graph")

fig.suptitle("Comparison of Line and Bar Charts")

plt.tight_layout()
plt.show()
```
