## Histograms in Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
x=np.random.randint(10,60,(50))
x
```

### Basic Histogram Syntax

```python
no=[20, 26, 55, 45, 32, 24, 25, 52, 57, 31, 59, 10, 22, 31, 25, 19, 14,
       19, 46, 38, 23, 11, 10, 20, 55, 30, 35, 20, 53, 35, 37, 44, 56, 42,
       43, 34, 59, 38, 36, 18, 56, 56, 52, 24, 37, 15, 45, 42, 34, 46]

plt.hist(no)
plt.show()
```

### Making Modifications in Histogram

```python
no=[20, 26, 55, 45, 32, 24, 25, 52, 57, 31, 59, 10, 22, 31, 25, 19, 14,
       19, 46, 38, 23, 11, 10, 20, 55, 30, 35, 20, 53, 35, 37, 44, 56, 42,
       43, 34, 59, 38, 36, 18, 56, 56, 52, 24, 37, 15, 45, 42, 34, 46]

plt.hist(no,edgecolor="black",linewidth=1)

plt.title("Batting Histogram Graph ")
plt.xlabel("Runs Scored")
plt.ylabel("Frequency Of Runs Scored")
plt.show()

```

### Making Use Of Bins Property to bucket the observations

```python
no=[20, 26, 55, 45, 32, 24, 25, 52, 57, 31, 59, 10, 22, 31, 25, 19, 14,
       19, 46, 38, 23, 11, 10, 20, 55, 30, 35, 20, 53, 35, 37, 44, 56, 42,
       43, 34, 59, 38, 36, 18, 56, 56, 52, 24, 37, 15, 45, 42, 34, 46]

bins=[10,20,30,40,50,60]
plt.hist(no,edgecolor="black",bins=bins,linewidth=2)

plt.title("Batting Histogram Graph ")
plt.xlabel("Runs Scored")
plt.ylabel("Frequency Of Runs Scored")
plt.show()
```

#### Using range parameter along with  autobins feature

```python
no=[20, 26, 55, 45, 32, 24, 25, 52, 57, 31, 59, 10, 22, 31, 25, 19, 14,
       19, 46, 38, 23, 11, 10, 20, 55, 30, 35, 20, 53, 35, 37, 44, 56, 42,
       43, 34, 59, 38, 36, 18, 56, 56, 52, 24, 37, 15, 45, 42, 34, 46]

bins=[10,20,30,40,50,60]
plt.hist(no,edgecolor="black",bins="auto",range=(0,100),linewidth=2)

plt.title("Batting Histogram Graph ")
plt.xlabel("Runs Scored")
plt.ylabel("Frequency Of Runs Scored")
plt.show()
```

#### Cumulative and Bottom Parameters

```python
no=[20, 26, 55, 45, 32, 24, 25, 52, 57, 31, 59, 10, 22, 31, 25, 19, 14,
       19, 46, 38, 23, 11, 10, 20, 55, 30, 35, 20, 53, 35, 37, 44, 56, 42,
       43, 34, 59, 38, 36, 18, 56, 56, 52, 24, 37, 15, 45, 42, 34, 46]

bins=[10,20,30,40,50,60]
plt.hist(no,edgecolor="black",bins="auto",linewidth=2,cumulative=-1)

plt.title("Batting Histogram Graph ")
plt.xlabel("Runs Scored")
plt.ylabel("Frequency Of Runs Scored")
plt.show()
```

```python
no=[20, 26, 55, 45, 32, 24, 25, 52, 57, 31, 59, 10, 22, 31, 25, 19, 14,
       19, 46, 38, 23, 11, 10, 20, 55, 30, 35, 20, 53, 35, 37, 44, 56, 42,
       43, 34, 59, 38, 36, 18, 56, 56, 52, 24, 37, 15, 45, 42, 34, 46]

bins=[10,20,30,40,50,60]
plt.hist(no,edgecolor="black",bins="auto",linewidth=2,bottom=10)

plt.title("Batting Histogram Graph ")
plt.xlabel("Runs Scored")
plt.ylabel("Frequency Of Runs Scored")
plt.show()
```
