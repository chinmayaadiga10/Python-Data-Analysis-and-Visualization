## Pie Charts in Matplotlib

```python
import matplotlib.pyplot as plt
```

### Basic Syntax

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]

plt.pie(x)
plt.show()
```

#### Adding labels to the pie chart 

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]

plt.pie(x,labels=y)
plt.show()
```

#### Understanding Explode Functions

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
explode=[0.0,0.0,0.2,0.0]

plt.pie(x,labels=y,explode=explode)# multiple values of x can have explode
plt.show()
```

#### Replacing default colors with custom colors

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]

plt.pie(x,labels=y,colors=color)
plt.show()
```

### Displaying percentages on the Pie Chart

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]

plt.pie(x,labels=y,colors=color,autopct="%1.1f%%",shadow=True)
plt.show()
```

### Radius and label-distance parameters

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]

plt.pie(x,labels=y,colors=color,autopct="%1.1f%%",radius=1.5)
plt.show()
```

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]

plt.pie(x,labels=y,colors=color,autopct="%1.1f%%",radius=0.75)
plt.show()
```

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]

plt.pie(x,labels=y,colors=color,autopct="%1.1f%%",labeldistance=1.3)
plt.show()
```

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]

plt.pie(x,labels=y,colors=color,autopct="%1.1f%%",labeldistance=0.8)
plt.show()
```

### Angle and Font size parameters 

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]
explode=[0.0,0.0,0.2,0.0]

plt.pie(x,labels=y,colors=color,explode=explode, autopct="%1.1f%%",labeldistance=1.1,startangle=0)
plt.show()
```

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]
explode=[0.0,0.0,0.2,0.0]

plt.pie(x,labels=y,colors=color,explode=explode, autopct="%1.1f%%",labeldistance=1.1,startangle=90)
plt.show()
```

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]
explode=[0.0,0.0,0.2,0.0]

plt.pie(x,labels=y,colors=color,explode=explode, autopct="%1.1f%%",labeldistance=1.1,startangle=180)
plt.show()
```

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]
explode=[0.0,0.0,0.2,0.0]

plt.pie(x,labels=y,colors=color,explode=explode, autopct="%1.1f%%",labeldistance=1.1,startangle=0,textprops={"fontsize":15})
plt.show()
```

### Title and Legend parameters

```python
y=["C","C++","Python","Java"]
x=[15,25,39,31]
color=["red","blue","green","yellow"]


plt.pie(x,labels=y,colors=color,autopct="%1.1f%%",labeldistance=1.1)
plt.title("Popularity of Programming Languages")
plt.legend(loc=4)
plt.show()
```

### Dot Pie Chart

```python
plt.pie([1])
plt.show()
```

### Pie Chart inside Pie Chart - Multiple Pie Chart

```python
y=["C","C++","Python","Java"]
x1=[15,25,39,31]
x2=[10,40,30,20]
color=["red","blue","green","yellow"]


plt.pie(x1,labels=y,autopct="%1.1f%%",labeldistance=1.1,radius=1.5)
plt.pie(x2,colors=color,autopct="%1.1f%%",labeldistance=1.1,radius=0.7)
plt.title("Popularity of Programming Languages")

plt.show()
```

### Doughnut Pie Chart

```python
y=["C","C++","Python","Java"]
x1=[15,25,39,31]
x2=[10,40,30,20]
color=["red","blue","green","yellow"]


plt.pie(x1,labels=y,autopct="%1.1f%%",labeldistance=1.1,radius=1.5)
plt.pie([1],colors="white",radius=1)
plt.title("Popularity of Programming Languages",loc="center")

plt.show()
```
