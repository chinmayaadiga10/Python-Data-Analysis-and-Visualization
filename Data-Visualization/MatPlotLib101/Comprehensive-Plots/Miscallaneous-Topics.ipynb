```python
import matplotlib.pyplot as plt
import numpy as np
```

## Fill between Plots

```python
x=[1,2,3,4,5]
y=[1,2,3,4,5]

plt.fill_between(x,y)

plt.plot(x,y)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,2,3,4,5]

plt.fill_between(x=[2,4],y1=2,y2=4)

plt.plot(x,y,color="black")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,2,3,4,5]

plt.fill_between(x,y,color="orange")

plt.plot(x,y,color="black")
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([1,2,3,4,5])

plt.fill_between(x,y,color="aqua",where=(x>=2)&(x<=4),alpha=0.4)

plt.plot(x,y,color="black")
plt.show()
```

## Subplots

```python
x=[1,2,3,4]
y=[1,2,3,4]

plt.subplot(2,2,1)
plt.plot(x,y,color="red")

plt.subplot(2,2,2)
plt.pie([1])

x1=[10,20,30,40]
plt.subplot(2,2,3)
plt.pie(x1)


x2=['a','s','d','f']
y2=[10,20,30,40]

plt.subplot(2,2,4)
plt.bar(x,y)

plt.show()
```

### Saving Figures

#### Creating Graph

```python
x=[1,2,3,4,5]
y=[2,4,3,6,4]

plt.plot(x,y,color="red")
plt.show()
```

#### Saving Graph

```python
x=[1,2,3,4,5]
y=[2,4,3,6,4]

plt.plot(x,y,color="red")
plt.savefig("line1.pdf",dpi=500,bbox_inches="tight")
plt.show()
```

### Axis in Matplotlib

```python
x=[1,2,3,4,5]
y=[3,1,5,2,4]

plt.plot(x,y)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[3,1,5,2,4]

plt.plot(x,y)
plt.xticks(x)
plt.yticks(x)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[3,1,5,2,4]

plt.plot(x,y)
plt.xticks(x,labels=["C++","C","Python","Java","JavaScript"])
plt.yticks(x)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[3,1,5,2,4]

plt.plot(x,y)
plt.xlim(0,10)
plt.ylim(0,10)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[3,1,5,2,4]

plt.plot(x,y)

plt.axis([0,10,0,7])
plt.show()
```

### Working with text in matplotlib

```python
x=[1,2,3,4,5]
y=[3,1,5,2,6]

plt.plot(x,y,color="red",label="trend")

plt.title("Arsenal Popularity",fontsize=20)
plt.xlabel("Years",fontsize=15)
plt.ylabel("Fan Following",fontsize=15)

plt.text(2.8,5.2,"2023",fontsize=15,style="italic")

plt.legend()

plt.show()
```
