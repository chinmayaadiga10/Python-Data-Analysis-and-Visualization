## Stem Plots

#### Used significantly in signals and electronic systems

```python
import matplotlib.pyplot as plt
```

### Basic syntax 

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y)
plt.show()
```

### Line,marker and bottom parameters

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt="--")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt="-.")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",markerfmt="r+")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",markerfmt="rd")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",markerfmt="rs")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",markerfmt="b*")
plt.show()
```

### Bottom parameter -> if 2, line starts at y=2

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",bottom=1)
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",bottom=3)
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",bottom=7)
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",bottom=1,basefmt="orange")
plt.show()
```

### Legend and Labels

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",bottom=1,label="Python")
plt.legend()
plt.show()
```

### Changing Orientation of Plot

```python
x=[1,2,3,4,5,6,7]
y=[6,3,4,7,6,4,2]
plt.stem(x,y,linefmt=":",orientation="horizontal")
plt.show()
```
