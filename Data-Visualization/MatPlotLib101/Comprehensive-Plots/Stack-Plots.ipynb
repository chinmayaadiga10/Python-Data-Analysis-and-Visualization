## Stack Plots in MatPlotLib
##### also called as Area Plot

```python
import matplotlib.pyplot as plt
```

### Single Stack Plot Syntax

```python
x=[1,2,3,4,5]
area=[1,2,3,4,5]

plt.stackplot(x,area)
plt.show()
```

```python
x=[1,2,3,4,5]
area=[5,2,6,4,7]

plt.stackplot(x,area)
plt.show()
```

### Multiple StackPlots

```python
x=[1,2,3,4,5]
area1=[1,2,3,4,5]
area2=[5,7,1,7,9]
area3=[3,8,6,2,3]

plt.stackplot(x,area1,area2,area3)
plt.show()
```

### Adding labels and legends to the plot

```python
x=[1,2,3,4,5]
area1=[1,2,3,4,5]
area2=[5,7,1,7,9]
area3=[3,8,6,2,3]

labels=["area1","area2","area3"]
plt.stackplot(x,area1,area2,area3,labels=labels)

plt.legend(loc=2)
plt.show()
```

### Passing custom colors to the graph

```python
x=[1,2,3,4,5]
area1=[1,2,3,4,5]
area2=[5,7,1,7,9]
area3=[3,8,6,2,3]

labels=["area1","area2","area3"]
colors=["black","gray","blue"]

plt.stackplot(x,area1,area2,area3,labels=labels,colors=colors)


plt.title("Stack Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.legend(loc=2)
plt.show()
```
