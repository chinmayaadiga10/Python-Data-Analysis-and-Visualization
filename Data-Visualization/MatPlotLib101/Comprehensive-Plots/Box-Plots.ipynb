## Box Plots in Matplotlib

```python
import matplotlib.pyplot as plt
```

### Basic Syntax

```python
x=[10,20,30,40,50,60,70]
plt.boxplot(x)
plt.show()
```


### Notch Plot

```python
x=[10,20,30,40,50,60,70]
plt.boxplot(x,notch=True)
plt.show()
```

### Horizontal Box Plot

```python
x=[10,20,30,40,50,60,70]
plt.boxplot(x,vert=False)
plt.show()
```

### Modifying width of the boxplot

```python
x=[10,20,30,40,50,60,70]
plt.boxplot(x,widths=0.4)# default width is set to 0.2
plt.show()
```

### Adding labels, colors and means in the boxplot

```python
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True)
plt.show()
```

```python
x=[10,20,30,40,50,60,70]
plt.boxplot(x,tick_labels=["Python"],showmeans=True)
plt.show()
```

### Outlier Representation in Boxplot

```python
x=[10,20,30,40,50,60,70,120]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True)
plt.show()
```

```python
x=[10,20,30,40,50,60,70,120,200]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True)
plt.show()
```

```python
x=[10,20,30,40,50,60,70,120,130]
plt.boxplot(x,tick_labels=["Python"],patch_artist=True)
plt.show()
```

### Plotting Multiple Boxplots

```python
x1=[10,20,30,40,50,60,70]
x2=[10,20,40,50,30,60,90]
y=[x1,x2]
plt.boxplot(y,tick_labels=["Python","C++"],patch_artist=True)
plt.show()
```
