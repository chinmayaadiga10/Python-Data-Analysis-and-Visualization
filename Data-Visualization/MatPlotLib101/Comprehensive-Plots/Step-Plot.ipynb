## Step Plots in Matplotlib


```python
import matplotlib.pyplot as plt
```

### Basic Syntax

```python
x=[1,2,3,4,5]
y=[1,2,3,4,5]

plt.step(x,y)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,2,3,5,4]

plt.step(x,y)
plt.show()
```

### Marker parameters 

```python
x=[1,2,3,4,5]
y=[2,2,3,5,4]

plt.step(x,y,marker="o",color="orange",mfc="blue",ms=8)
plt.show()
```

### Adding labels and legends

```python
x=[1,2,3,4,5]
y=[2,2,3,5,4]

plt.step(x,y,marker="o",color="orange",mfc="blue",ms=8,label="python")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid()
plt.legend()
plt.show()
```
