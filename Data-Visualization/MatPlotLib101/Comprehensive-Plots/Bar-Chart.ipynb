## Bar Plots

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Basic Bar Plot

```python
x=["Python","C++","C","Java"]
y=[85,70,65,92]
plt.bar(x,y)
plt.show()
```

### Adding labels, titles, colors, fontsize and other properties

```python
x=["Python","C++","C","Java"]
y=[85,70,65,92]
plt.bar(x,y,width=0.45,color="orange")
plt.xlabel("Programming Language",)
plt.ylabel("Popularity")
plt.title("Programming Language v/s Popularity")
plt.show()
```

#### Changing text alignment, adding individual colors

```python
x=["Python","C++","C","Java"]
y=[85,70,65,92]
colors=["Red","Yellow","Blue","Orange"]
plt.bar(x,y,width=0.45,color=colors,align="edge",edgecolor="black",linewidth=2,linestyle=":")# align is set to center by default
plt.xlabel("Programming Language",)
plt.ylabel("Popularity")
plt.title("Programming Language v/s Popularity")
plt.show()
```

#### Adding labels and legends to the graph

```python
x=["Python","C++","C","Java"]
y=[85,70,65,92]
plt.bar(x,y,width=0.45,color="green",label="Popularity",alpha=0.4)#
plt.xlabel("Programming Language",)
plt.ylabel("Popularity")
plt.title("Programming Language v/s Popularity")
plt.legend()
plt.show()
```

### Creating Double Bar Graph

#### Overlapped Double Bar Graph

```python
x=["Python","C++","C","Java"]
y=[85,70,65,92]
z=[20,30,40,50]
plt.bar(x,y,width=0.45,color="orange",label="Popularity-1")
plt.bar(x,z,width=0.45,color="blue",label="Popularity-2")
plt.xlabel("Programming Language",)
plt.ylabel("Popularity")
plt.title("Programming Language v/s Popularity")
plt.legend()
plt.show()
```

#### Adjacent Double Bar Graph

```python
x=["Python","C++","C","Java"]
y=[85,70,65,92]
z=[20,30,40,50]

width=0.4

p=np.arange(len(x))
p1=[j+width for j in p]

plt.xlabel("Programming Language",)
plt.ylabel("Popularity")
plt.title("Programming Language v/s Popularity")

plt.bar(p,y,width=0.4,color="orange",label="Popularity-1")
plt.bar(p1,z,width=0.4,color="blue",label="Popularity-2")

plt.xticks(p+width/2,x,rotation=10)
plt.legend()
plt.show()
```
