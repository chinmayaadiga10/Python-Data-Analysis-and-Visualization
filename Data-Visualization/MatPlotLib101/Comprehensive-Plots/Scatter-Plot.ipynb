## Scatter Plots in MatPlotLib

```python
import matplotlib.pyplot as plt
```

### Basic Syntax 

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
plt.scatter(x,y)
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
plt.scatter(x,y,color="red")
plt.title("Scatter Plot",fontsize=10)
plt.xlabel("Day")
plt.ylabel("No.of Times")
plt.show()
```

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
colors=["red","yellow","orange","green","blue","black","pink"]
plt.scatter(x,y,color=colors)
plt.title("Scatter Plot",fontsize=10)
plt.xlabel("Day")
plt.ylabel("No.of Times")
plt.show()
```

#### Changing size of each observation

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
colors=["red","yellow","orange","green","blue","black","pink"]
sizes=[10,5,6,19,4,7,3]
plt.scatter(x,y,color=colors,s=sizes)
plt.title("Scatter Plot",fontsize=10)
plt.xlabel("Day")
plt.ylabel("No.of Times")
plt.show()
```

#### Changing opacity using alpha property

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
colors=["red","yellow","orange","green","blue","black","pink"]
plt.scatter(x,y,color=colors,alpha=0.4)
plt.title("Scatter Plot",fontsize=10)
plt.xlabel("Day")
plt.ylabel("No.of Times")
plt.show()
```

#### Playing with different styles of recording observation

```python
x=[1,2,3,4,5,6,7]
y=[2,3,1,4,5,3,6]
colors=["red","yellow","orange","green","blue","black","pink"]
sizes=[100,200,100,300,400,500,200]
plt.scatter(x,y,color=colors,s=sizes,marker="*",edgecolors="black",linewidths=1)
plt.title("Scatter Plot",fontsize=10)
plt.xlabel("Day")
plt.ylabel("No.of Times")
plt.show()
```
