```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("./netflix_titles.csv")
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=["type","release_year","rating","country","duration"])
```

```python
df=df.dropna(subset=["type","release_year","rating","country","duration"])
```

```python
type_counts=df['type'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(type_counts.index,type_counts.values,color=["skyblue","orange"])
plt.title("No.of Movies v/s TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("movies_vs_tv_shows.png",dpi=500,bbox_inches="tight")
plt.show()
```

```python
rating_counts=df['rating'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(rating_counts,labels=rating_counts.index,autopct="%1.1f%%",startangle=90)
plt.title("Percentage of content ratings")
plt.tight_layout()
plt.savefig("content_ratings_pie.png",dpi=500,bbox_inches="tight")
plt.show()

```

### Movie Duration Distribution using Histogram

```python
movie_df=df[df['type']=="Movie"].copy()
movie_df
```

```python
movie_df['duration_int']=movie_df['duration'].str.replace(" min","").astype(int)
```

```python
movie_df
```

```python
plt.figure(figsize=(8,6))
plt.hist(movie_df['duration_int'],bins=30,color="purple",edgecolor="black")
plt.title("Distribution of movie duration")
plt.xlabel("Duration (minutes)")
plt.ylabel("No.of movies")
plt.tight_layout()
plt.savefig("movie_duration_histogram.png",dpi=500,bbox_inches="tight")
plt.show()
```

```python
release_counts=df['release_year'].value_counts().sort_index()
plt.figure(figsize=(8,6))
plt.scatter(release_counts.index,release_counts.values,color="orange",label="no.of movies")
plt.title("No.of shows released in particular year")
plt.xlabel("Year")
plt.ylabel("No.of shows released")
plt.legend()
plt.savefig("no_of_shows_released.png",dpi=500,bbox_inches="tight")
plt.tight_layout()
plt.show()
```
