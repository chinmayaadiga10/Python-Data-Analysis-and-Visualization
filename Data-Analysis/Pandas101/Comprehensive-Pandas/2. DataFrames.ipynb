```python
import numpy as np
import pandas as pd
```

### Creating Dataframe from dictionary and Array

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
```

```python
# in case of dictionary, the key acts as the column names
pd.DataFrame(data)
```

```python
#in case of array, column names have to be explicitly mentioned
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
```

```python
pd.DataFrame(data_list)
```

```python
# specifying explicit column names in case of list
columns=["Name","Age","City","Salary"]
pd.DataFrame(data_list,columns=columns)
```

```python
# storing dataframe in 'df' variable
df=pd.DataFrame(data)
```

```python
df
```

#### Creating new column

```python
df['Designation']=["Engineer","Doctor","Engineer","Doctor"]
#here, designation is the column name and the values inside the list appear in the column. each value in a row
```

```python
df
```

#### Removing Columns

```python
# by default axis =0, (rows), we have to change it to 1 to delete all the values from the column
df.drop('Designation',axis=1)
```

```python
# we can notice that the designation is not dropped, but just the copy was shown in the above result
df
```

```python
# by using inplace, permanent changes take place in the dataframe
df.drop('Designation',axis=1,inplace=True)
```

```python
df
```

#### Removing first row by selecting index and setting axis=0

```python
df.drop(0,axis=0)
# axis 0 targets the rows
```
