## Creating Series in Pandas

#### Series is simply a single column in pandas. Multiple series are combined to form a DataFrame

```python
import numpy as np
import pandas as pd
```

```python
labels=['a','b','c','d','e']
my_list=[11,22,33,44,55]
dictionary={1:10,2:20,3:30}
arr=np.array([10,20,30,40])
```

#### Making use of List to create Series

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index=labels)
#here, index is set to labels, which means 0,1 ... will be replaced by the labels list.
#however, the length of the labels list should be same as the length of the series
```

```python
pd.Series(dictionary)
# in dictionary, the keys become the indices and the values become the series values
```

```python
new_dict={'a':10,'b':20,'c':30}
pd.Series(new_dict)
```

```python
pd.Series(arr)
```
