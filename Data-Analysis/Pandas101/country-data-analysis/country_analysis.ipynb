### Finding all countries which have ***Republic*** in the name

```python
def check_republic(txt):
    txt=txt.lower()
    if 'republic' in txt:
        return True
    return False
```

```python
check_republic(" People's Republic of Angola")
```

```python
check_republic("Islamic State of Afghanistan")
```

```python
df['country_long']
```

```python
df['country_long'].apply(check_republic)
```

```python
df[df['country_long'].apply(check_republic)]
```

```python
df[df['country_long'].apply(check_republic)].count()
```
