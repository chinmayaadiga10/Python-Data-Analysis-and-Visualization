### Vector Operations in NumPy

```python
import numpy as np
```

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])

#Vector addition -> 
print(f"Vector addition : {vector1+vector2}")


#Vector multiplication ->
print(f"Vector multiplication : {vector1*vector2}")

# Dot product in vectors ->
print(f"Vector dot product is {np.dot(vector1,vector2)}")


#Finding the angle between two vectors -> 
angle=np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print(f"The angle between the two vectors is : {angle}")

```

##### Vectorized Operations

```python
# the operations that are performed individually on each element of the array
restaurant_types=np.array(['cafe','chat point','chinese','pizza','fine dine'])

vectorized_upper=np.vectorize(str.upper)#vectorising normal python method
print(f"Vectorized upper : {vectorized_upper(restaurant_types)}")
```
