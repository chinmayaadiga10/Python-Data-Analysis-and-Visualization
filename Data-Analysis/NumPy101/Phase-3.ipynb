### Business operations simulation with NumPy

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Kapoor's Cafe
    [2, 120000, 140000, 160000, 190000],  # SLV Coffee Bar
    [3, 200000, 230000, 260000, 300000],  # Kedia's fun food
    [4, 180000, 210000, 240000, 270000],  # Corner House
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
print("=== Zomato Sales Data Analysis")
print(f"Shape of dataset :{sales_data.shape}")

#Returning sales data of the first 3 restaurants -> 
print(f"Data of the first 3 restaurants : \n{sales_data[0:3]}")
```

Finding total sales of all restaurants across years

```python
print(np.sum(sales_data,axis=0))

yearly_total=np.sum(sales_data[:,1:],axis=0)
print(yearly_total)
```

Finding minimum sales per restaurant

```python
min_sales=np.min(sales_data[:,1:],axis=1)
print(min_sales)
```

Maximum sales per year

```python
max_sales=np.max(sales_data[:,1:],axis=0)
print(max_sales)
```

Average sales per restaurant

```python
avg_sales=np.mean(sales_data[:,1:],axis=1)
print(avg_sales)
```

Finding Cumulative Sales

```python
cumulative_sales=np.cumsum(sales_data[:,1:],axis=1)
print(cumulative_sales)

plt.figure(figsize=(5,3))
plt.plot(np.mean(cumulative_sales,axis=0))
plt.title("Average cumulative sales per year across all restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
